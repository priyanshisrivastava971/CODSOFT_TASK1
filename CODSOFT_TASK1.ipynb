```python
import gradio as gr
import random
import re
from datetime import datetime


class AdvancedChatbot:

    def __init__(self):
        self.bot_name = "CodSoft AI Assistant"

        self.responses = {
            "greeting": [
                "Hello! Welcome. How can I help you?",
                "Hey! Nice to see you. What's on your mind?",
                "Hi there! How may I assist you today?"
            ],
            "thanks": [
                "You're most welcome!",
                "Happy to help you!",
                "Anytime! Let me know if you need anything else."
            ],
            "study": [
                "Make a daily study schedule and practice consistently.",
                "Break difficult topics into smaller parts.",
                "Practice coding regularly to improve your skills."
            ],
            "career": [
                "Build practical projects and upload them to GitHub.",
                "Improve your technical and communication skills.",
                "Start applying for internships and build your portfolio."
            ],
            "goodbye": [
                "Goodbye! Have a wonderful day!",
                "See you soon. Take care!",
                "It was nice talking to you!"
            ]
        }

    def get_response(self, message):

        message = message.lower().strip()

        if not message:
            return "Please enter a message."

        if re.search(r"\b(hi|hello|hey)\b", message):
            return random.choice(self.responses["greeting"])

        if "your name" in message or "who are you" in message:
            return (
                "I am CodSoft AI Assistant, "
                "a rule-based chatbot developed using Python."
            )

        if "date" in message or "today" in message:
            return datetime.now().strftime(
                "Today's date is %d %B %Y."
            )

        if "time" in message:
            return datetime.now().strftime(
                "The current time is %I:%M %p."
            )

        if "how are you" in message:
            return (
                "I am working perfectly! Thank you for asking. "
                "How can I help you?"
            )

        if any(word in message for word in [
            "study", "exam", "learning", "college"
        ]):
            return random.choice(self.responses["study"])

        if any(word in message for word in [
            "career", "job", "internship", "resume"
        ]):
            return random.choice(self.responses["career"])

        match = re.search(
            r"(-?\d+(?:\.\d+)?)\s*([+\-*/])\s*(-?\d+(?:\.\d+)?)",
            message
        )

        if match:
            num1 = float(match.group(1))
            operator = match.group(2)
            num2 = float(match.group(3))

            if operator == "+":
                answer = num1 + num2
            elif operator == "-":
                answer = num1 - num2
            elif operator == "*":
                answer = num1 * num2
            elif operator == "/":
                if num2 == 0:
                    return "Division by zero is not allowed."
                answer = num1 / num2

            return f"The answer is **{answer:g}**."

        if "thank" in message:
            return random.choice(self.responses["thanks"])

        if re.search(r"\b(bye|goodbye|exit)\b", message):
            return random.choice(self.responses["goodbye"])

        if "what can you do" in message or "help" in message:
            return """
            I can help you with:

            1. Greetings and conversation
            2. Study suggestions
            3. Career guidance
            4. Current date and time
            5. Basic mathematical calculations
            6. General predefined responses

            Try asking me something!
            """

        if "what happened" in message:
            return "Tell me what happened. I'm listening!"

        return (
            "I'm not sure I understand your question yet. "
            "Try asking about studies, career, date, time "
            "or mathematical calculations."
        )


bot = AdvancedChatbot()


def chat(message, history):

    if not message.strip():
        return history, ""

    response = bot.get_response(message)

    history = history or []

    history.append({
        "role": "user",
        "content": message
    })

    history.append({
        "role": "assistant",
        "content": response
    })

    return history, ""


def clear_chat():
    return [], ""


custom_css = """
.gradio-container {
    max-width: 1200px !important;
    margin: auto !important;
}

#main-title {
    text-align: center;
    padding: 15px;
}

#chat-area {
    border-radius: 15px;
}

#send-button {
    background: #2563eb;
    color: white;
}

#sidebar {
    border-right: 1px solid var(--border-color-primary);
    padding: 15px;
}

footer {
    display: none !important;
}
"""

with gr.Blocks(title="CodSoft AI Chatbot") as demo:

    gr.Markdown(
        """
        # 🤖 CodSoft AI Assistant
        ### Your Personal Rule-Based Chatbot
        """,
        elem_id="main-title"
    )

    with gr.Row():

        with gr.Column(scale=1, min_width=180, elem_id="sidebar"):

            gr.Markdown("## 💬 Chat Menu")

            new_chat = gr.Button(
                "＋ New Chat",
                variant="primary"
            )

            gr.Markdown(
                """
                ### About

                A Python-based chatbot
                using predefined rules
                and pattern matching.

                ### Features

                - Smart responses
                - Calculator
                - Study guidance
                - Career guidance
                - Date and time
                """
            )

        with gr.Column(scale=4):

            chatbot = gr.Chatbot(
                label="Conversation",
                height=480,
                show_label=False,
                placeholder="👋 Start a conversation with CodSoft AI!"
            )

            with gr.Row():

                message = gr.Textbox(
                    placeholder="Message CodSoft AI...",
                    show_label=False,
                    scale=8,
                    lines=1
                )

                send_button = gr.Button(
                    "Send ➤",
                    variant="primary",
                    scale=1,
                    elem_id="send-button"
                )

            gr.Markdown(
                "*CodSoft AI Assistant | Python Rule-Based Chatbot*"
            )

    send_button.click(
        chat,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

    message.submit(
        chat,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

    new_chat.click(
        clear_chat,
        inputs=[],
        outputs=[chatbot, message]
    )

demo.launch(
    theme=gr.themes.Soft(primary_hue="blue", neutral_hue="slate"),
    css=custom_css,
    share=True,
    debug=True
)
```
